## Min Max Scallilng 
* Rescale the value to the range of 0 , 1 
* All the features need to be scaled, especially for algorithm using distance metric

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
```

```python
data = {
    'Name': ["Abdullah","Omer","Talha","Ali"],
    "Age":[18,25,30,50]
}
df = pd.DataFrame(data)
df
```

```python
scaler = MinMaxScaler()
df['Normalized Age'] = scaler.fit_transform(df[['Age']])
df
```
