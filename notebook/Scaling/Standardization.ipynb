```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
```

```python
data = {
    'Name': ["Abdullah", "Omer", "Talha", "Ali"],
    "Age": [18, 25, 30, 50]
}
df = pd.DataFrame(data)
df
```

```python
# Step 1: Get age list
ages = data['Age']

# Step 2: Calculate mean
mean_age = sum(ages) / len(ages)

# Step 3: Calculate standard deviation (sample or population – using population here)
squared_diffs = [(x - mean_age) ** 2 for x in ages]
std_age = (sum(squared_diffs) / len(ages)) ** 0.5

# Step 4: Apply scaling formula x_scaled = (x - mean) / std
scaled_ages = [(x - mean_age) / std_age for x in ages]

# Output
print(f"Mean: {mean_age}")
print(f"Standard Deviation: {std_age}")
print("Scaled Ages:")
for original, scaled in zip(ages, scaled_ages):
    print(f"Original: {original} → Scaled: {scaled}")
```

```python
import numpy as np

ages = np.array(data['Age'])
mean_age = np.mean(ages)
std_age = np.std(ages)

scaled_ages = (ages - mean_age) / std_age
print("Scaled ages:", scaled_ages)
```

```python
scaler = StandardScaler()
df['Standard Scaled'] = scaler.fit_transform(df[['Age']])
df
```
