* Using KNeighborsClassifier to predict whether a customer will buy or not based on [Age, Income].
* First, training a model without any feature scaling.
* Then training models with MinMaxScaler and StandardScaler.
* Predicting for a new customer: [40, 75000] using all three models (unscaled, min-max scaled, standard scaled).
* Comparing the results.

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.neighbors import KNeighborsClassifier
import numpy as np
```

```python
# 🧪 Sample data: [Age, Income]
X = np.array([
    [25, 50000],   # Customer A - 0
    [30, 60000],   # Customer B - 0 
    [45, 80000],   # Customer C - 1
    [35, 20000],   # Customer D - 0
    [50, 100000],  # Customer E - 1
])
```

```python
# Labels (e.g. 0 = won't buy, 1 = will buy)
y = np.array([0,0,1,0,1])
new_customer = np.array([[28, 55000]])
```

```python
# 1. Train KNN WITHOUT Scaling
# ---------------------------------
model_unscaled = KNeighborsClassifier(n_neighbors=3)
model_unscaled.fit(X, y)
prediction_unscaled = model_unscaled.predict(new_customer)
```

```python
# 2. MinMax Scaling
# ---------------------------------
minmax_scaler = MinMaxScaler()
X_minmax_scaled = minmax_scaler.fit_transform(X)
new_customer_minmax = minmax_scaler.transform(new_customer)
```

```python
model_minmax = KNeighborsClassifier(n_neighbors=3)
model_minmax.fit(X_minmax_scaled, y)
prediction_minmax = model_minmax.predict(new_customer_minmax)
```

```python
# 3. Standard Scaling
# ---------------------------------
standard_scaler = StandardScaler()
X_standard_scaled = standard_scaler.fit_transform(X)
new_customer_standard = standard_scaler.transform(new_customer)
```

```python
model_standard = KNeighborsClassifier(n_neighbors=3)
model_standard.fit(X_standard_scaled, y)
prediction_standard = model_standard.predict(new_customer_standard)
```

```python
# 📤 Print Results
# ---------------------------------
print("Original Data:\n", X)
print("------------------------------")
print("New Customer (Original):", new_customer)
print("------------------------------")
print("Prediction WITHOUT scaling:", prediction_unscaled)
print("------------------------------")
print("Prediction WITH MinMax scaling:", prediction_minmax)
print("------------------------------")
print("Prediction WITH Standard scaling:", prediction_standard)
print("------------------------------")
print("New Customer (MinMax Scaled):", new_customer_minmax)
print("------------------------------")
print("New Customer (Standard Scaled):", new_customer_standard)
```

```python
# ----------------------------
# 🔎 Try a few new customers
# ----------------------------
test_customers = np.array([
    [28, 55000],   # Likely 0
    [48, 95000],   # Likely 1
    [34, 25000],   # Likely 0
])

print("\n🧪 Testing New Customers:")
for i, cust in enumerate(test_customers, start=1):
    scaled_cust_minmax = minmax_scaler.transform([cust])
    scaled_cust_standard = standard_scaler.transform([cust])
    
    pred_minmax = model_minmax.predict(scaled_cust_minmax)[0]
    pred_standard = model_standard.predict(scaled_cust_standard)[0]
    
    print(f"Customer {i} ({cust}):")
    print(f"  🟦 MinMax prediction: {pred_minmax}")
    print(f"  🟥 Standard prediction: {pred_standard}")
```
