## One-Hot Encoding
* Converts each category into a separate column with 0 or 1 (dummy variables).
* Best for nominal (unordered) data.

```python
import pandas as pd
```

```python
data = {
    'color':['Red','Green','Blue']
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df_encoded = pd.get_dummies(df['color'])
```
