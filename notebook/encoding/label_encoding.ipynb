# Label Encoding
* Assign a each category unique no
* Used for ordinal or no ordered data
* It may imply order to the nominal data

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
```

```python
data  = {'color':['Red','Blue',"Green","Yellow"]}
df = pd.DataFrame(data)
```

```python
df
```

```python
le = LabelEncoder()
df['Encoded_age'] = le.fit_transform(df[['color']])
```
