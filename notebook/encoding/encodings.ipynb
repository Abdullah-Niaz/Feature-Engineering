# Label Encoder

```python
from sklearn.preprocessing import LabelEncoder
```

```python
data = ['Low', 'Medium','High','Medium','Low']
```

```python
le = LabelEncoder()
```

```python
ecnoded_data = le.fit_transform(data)
```

```python
print(f"Encoded Data: {ecoded_data}")
```

# One - Hot Encoding

```python
# from sklearn.preprocessing import LablelEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
data = [['low'], ['Medium'],['High'],['Medium'],['low']]
```

```python
encoder = OrdinalEncoder(categories=[['low', 'Medium', 'High']])
ecnoded_data = encoder.fit_transform(data)
```

```python
print(f"Encoded Ordinal Data: {ecnoded_data}")
```
